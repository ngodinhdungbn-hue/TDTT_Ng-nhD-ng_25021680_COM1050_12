```python
def is_Happy_Value(n):
    try:
        n=list(map(int,str(n)))
        m=0
        for i in n:
            m=m+i**2
        if m==1:
            print('YES')
        else:
            is_Happy_Value(m)
    except RecursionError:
        print('NO')


try:
    n=int(input('Nhập'))
except ValueError as e:
        print(f"ValueError : {e}")

is_Happy_Value(n)
```

```python
try:
    n=int(input())
    res=[]
    d={}
    while n>0:
        x=int(input())
        res.append(x)
        n-=1
    for x in res:
        if x not in d:
            d[x]=1
        else:
            d[x]+=1
    max=0
    for x in d:
        if d[x]>max:
            max=d[x]
    if max==1:
        sum=0
        for i in res:
            sum+=i
        print(sum)
    else:
        raise ValueError
except ValueError:
    print("Mang khong hop le")
```

```python
def is_valid_regex(s):
    operators = {'*', '+', '?'}
    n = len(s)
    i = 0

    while i < n:
        if s[i] == '\\':
            if i == n - 1:
                return False
            i += 2
            continue

        if s[i] in operators:
            if i == 0:
                return False
            if s[i - 1] in operators:
                return False
        i += 1
    return True


t = int(input())
for _ in range(t):
    s = input().strip()
    print("True" if is_valid_regex(s) else "False")
```
